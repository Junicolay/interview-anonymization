## Future Work — Fine-Tuning Strategies

After validating the metric-guided revision baseline, we will investigate whether fine-tuning can teach the model to generate content-preserving paraphrases more reliably, reducing the need for repeated LLM calls.

Three approaches will be considered.

### 1. Prompt Engineering and Metric-Guided Revision

The initial baseline keeps the LLM parameters frozen. Content-preservation metrics are used at inference time to evaluate generated paraphrases and trigger targeted revisions when necessary.

This approach establishes whether the proposed metrics and revision mechanism improve content preservation without modifying the underlying model.

### 2. Supervised Fine-Tuning (SFT) with LoRA

The second approach adapts an open-weight LLM using supervised examples of successful contextual paraphrases.

Each training example consists of an input containing the original transcript and relevant context, together with a target paraphrase that preserves the intended meaning while modifying linguistic style.

Semantic Similarity, TPI, and NLI can help filter and assess candidate training examples. However, metric scores alone will not be treated as sufficient evidence of paraphrase quality; a manually validated subset will be used to assess the reliability of this filtering process.

Low-Rank Adaptation (LoRA) will be considered to reduce the computational cost of fine-tuning by training a small set of additional parameters rather than updating all model parameters.

The objective is to teach the model to produce suitable paraphrases more consistently in a single generation step.

### 3. Preference Optimization with DPO

A third approach will investigate Direct Preference Optimization (DPO) after SFT.

For each original transcript, multiple paraphrase candidates can be generated and evaluated. These candidates will be used to construct preference pairs:

- **Preferred candidate:** Preserves the intended meaning and relevant information while providing sufficient stylistic modification.
- **Rejected candidate:** Introduces semantic distortions, loses important information, or fails to modify identifying linguistic characteristics adequately.

Semantic Similarity, TPI, and NLI can contribute to identifying candidate preferences, while privacy evaluation and human validation will help prevent misleading or contradictory metric signals.

DPO will then optimize the model to favor preferred outputs over rejected alternatives. Unlike the initial revision baseline, this approach changes the model parameters based on preference data.

### 4. Comparison of the Approaches

The three approaches will be compared under consistent experimental conditions.

| Approach | Model parameters updated? | Role of the metrics |
|---|---|---|
| Prompt engineering and metric-guided revision | No | Evaluate candidates and trigger revisions |
| SFT with LoRA | Yes | Help select and validate supervised training examples |
| SFT followed by DPO | Yes | Help construct preference pairs for further optimization |

The comparison will assess whether fine-tuning improves content preservation beyond what can be achieved through prompt engineering and metric-guided revision alone.

### 5. Evaluation Criteria

All approaches will be evaluated using the same content-preservation and privacy criteria:

- Greedy Alignment Score (GAS), where higher is better.
- DTW Similarity, where higher is better.
- Semantic Similarity and TPI.
- NLI entailment, neutral, and contradiction probabilities.
- Equal Error Rate (EER) of the content-based speaker re-identification attack.
- Computational cost, inference latency, and the number of required LLM calls.

The main objective is to determine whether fine-tuning can improve the trade-off between semantic preservation and privacy, rather than simply maximizing textual similarity.

### 6. Proposed Development Order

The development will follow a progressive strategy:

1. Implement and evaluate the metric-guided revision baseline.
2. Establish whether the metrics reliably identify meaningful content-preservation errors.
3. Build and validate a dataset of contextual paraphrases for SFT.
4. Fine-tune an open-weight model using LoRA.
5. Evaluate whether SFT improves GAS and DTW Similarity without degrading privacy.
6. If sufficient preference data can be constructed, apply DPO to further optimize the trade-off.
7. Compare all approaches against the original reference pipeline.

Fine-tuning will only be pursued if the baseline experiments provide evidence that the proposed metrics and revision strategy offer a reliable foundation for model optimization.

The eventual goal is to develop a content-aware anonymization model that preserves semantic information while modifying linguistic style sufficiently to reduce speaker re-identification risk.
